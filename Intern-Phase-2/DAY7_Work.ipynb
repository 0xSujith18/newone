```python
# Demonstrating keyword search vs Semantic search


# Selling collection of documents about data

documents = [
    "ETL is used to clean and transform data",
    "A vehicle is a mode of a transportation",
    "Cars and trucks are powerful automobiles",
    "SQL is used to query Databases",
    "Machine learning trains the model on data",
]

# Performing the keyword searching ..

keyword = "Vehicle"

print("="*80)
print(f"Keyword search for the : {keyword}")
print("="*80)

for i,doc in enumerate(documents):

  if keyword.lower() in doc.lower():
    print(f" Found [doc_{i}]:{doc}")
  else:
    print(f"Missed [doc_{i}:{doc}]")

print("\n=====================================================================================")
print("Problem : doc_2 talks about 'cars and trucks' which are vehicles ")
print("But the keyword search missed it")
print("====================================================================================")
```

```python
# Demonstrating keyword search vs Semantic search
# Selling collection of documents about data
# Failure examples where keywords mostly fails

failure_examples = [
    {"query":"I fell sick",  "misses":"I am unwell, patient was fever"},
    {"query":"How to cook rice", "misses":"steps to prepare"},

]

print("Keyword search failures:")
print(80*"=")

# Demonstrating how semantic search can win

print("\n" + 80*"=")
print("Semantic Search Wins:")
print(80*"=")

# New documents for semantic search demonstration
semantic_docs = [
    "I am feeling unwell today, I think I have a fever.",
    "Detailed instructions on how to cook perfect fluffy rice.",
    "A car is a type of vehicle with four wheels.",
    "Learning about machine intelligence models."
]

semantic_queries = [
    "I fell sick",
    "How to cook rice",
    "Vehicle information",
    "AI and learning models"
]

# Load a SentenceTransformer model for semantic search
# This model converts text into numerical vectors (embeddings)
model = SentenceTransformer('all-MiniLM-L6-v2')

# Encode documents and queries into embeddings
doc_embeddings = model.encode(semantic_docs)
query_embeddings = model.encode(semantic_queries)

# Perform semantic search for each query
for i, query in enumerate(semantic_queries):
    print(f"\nQuery: {query}")
    # Calculate cosine similarity between query and all documents
    # Cosine similarity measures the angle between two vectors, indicating how similar they are.
    similarities = np.dot(query_embeddings[i], doc_embeddings.T)

    # Find the index of the most similar document
    most_similar_doc_idx = np.argmax(similarities)

    print(f"  Most relevant document (Semantic Search): {semantic_docs[most_similar_doc_idx]} (Similarity: {similarities[most_similar_doc_idx]:.4f})")
    print(f"  Keyword search might miss: {failure_examples[i]['misses'] if i < len(failure_examples) else 'N/A'}")

print("\n" + 80*"=")
print("Semantic search successfully finds semantically similar content, overcoming keyword limitations.")
print(80*"=")
```

```python
# Generate the first embedding

sentence = "ETL is used to clean and transform data"

embedding = model.encode(sentence)

print(f"Input sentence : {sentence}")

print(f"Embedding type : {type(embedding)}")

print(f"Shape :{embedding.shape}")

sentences = [
    "ETL is used to clean and transorm data",
    "Data transformation is a key pipeline step",
    "The sky is blue and clouds are white ",
]

embeddings = model.encode(sentences)
print(f"Number of sentences : {len(sentences)}")
print(f"Embedding shape : {embeddings.shape}")
print()

print("Each sentence is mapped to a 384 dimensional vector")
print("Each sentence is one row's embedding :")
for i,sent in enumerate(sentences):
  print(f"Sentence {i}: shape={embeddings[i].shape}, first 5 values = {embeddings[i][:7].round(3)}")
```

```python
# Calcualting cosine similarity
def cosine_similarity(vec_a,vec_b):
  dot_product =np.dot(vec_a,vec_b)
  norm_a = np.linalg.norm(vec_a) # Magnitude ||A||
  norm_b = np.linalg.norm(vec_b)
  similarity = dot_product / (norm_a*norm_b)
  return similarity
# Compare the first sentence with the second
sim_0_1 = cosine_similarity(embeddings[0], embeddings[1])
print(f"Similarity between '{sentences[0]}' and '{sentences[1]}': {sim_0_1:.4f}")

# Compare the first sentence with the third
sim_0_2 = cosine_similarity(embeddings[0], embeddings[2])
print(f"Similarity between '{sentences[0]}' and '{sentences[2]}': {sim_0_2:.4f}")

# Compare the second sentence with the third
sim_1_2 = cosine_similarity(embeddings[1], embeddings[2])
print(f"Similarity between '{sentences[1]}' and '{sentences[2]}': {sim_1_2:.4f}")
```
